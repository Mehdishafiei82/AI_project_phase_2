# بررسی محیط Python برای سیستم توصیه‌گر کتاب

این نوت‌بوک نسخه‌های کتابخانه‌های موردنیاز پروژه (numpy، pandas، scikit-learn، matplotlib، seaborn) را چک می‌کند
و یک تست سریع cosine similarity اجرا می‌کند تا مطمئن شویم همه‌چیز درست نصب شده است.

- سنجه عملکرد (Performance Measure): مرتبط بودن پیشنهادها، تنوع، تازگی
- محیط (Environment): پروفایل کاربر، کاتالوگ کتاب، سیگنال‌های مشارکتی
- عملگرها (Actuators): نمایش لیست پیشنهاد، اعلان/ایمیل
- حسگرها (Sensors): کلیک، امتیاز، نظرات، جستجو

```python
import sys
import numpy
import pandas
import sklearn
import matplotlib
import seaborn

print("Python:", sys.version)
print("numpy:", numpy.__version__)
print("pandas:", pandas.__version__)
print("scikit-learn:", sklearn.__version__)
print("matplotlib:", matplotlib.__version__)
print("seaborn:", seaborn.__version__)
print("\nهمه کتابخانه‌ها با موفقیت ایمپورت شدند ✔")
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# بردار ویژگی دو کتاب فرضی (مثلاً بردار TF-IDF خلاصه کتاب‌ها)
book_1 = np.array([[1.0, 2.0, 3.0]])
book_2 = np.array([[2.0, 4.0, 6.0]])   # کاملاً هم‌جهت با کتاب ۱ → شباهت = 1
book_3 = np.array([[3.0, 0.0, 0.0]])   # جهت متفاوت

sim_12 = cosine_similarity(book_1, book_2)[0][0]
sim_13 = cosine_similarity(book_1, book_3)[0][0]

print(f"شباهت کسینوسی کتاب ۱ و کتاب ۲: {sim_12:.4f}")
print(f"شباهت کسینوسی کتاب ۱ و کتاب ۳: {sim_13:.4f}")
assert 0.99 < sim_12 <= 1.0, "خطا در محاسبه شباهت!"
print("\nتست cosine similarity موفق بود ✔")
```
